```python
import pandas as pd
from itertools import combinations
from collections import defaultdict

# 1. Lê o CSV pulando a linha 1 e usando a linha 2 como cabeçalho
df = pd.read_csv('/content/Relação das características', header=1)

# 2. Remove linhas onde o ID ou o Eixo interpretativo são nulos
df_limpo = df.dropna(subset=['ID', 'Eixo interpretativo']).copy()

# 3. Remove espaços extras nos nomes dos eixos para evitar duplicidades por digitação
df_limpo['Eixo interpretativo'] = df_limpo['Eixo interpretativo'].str.strip()

# 4. Agrupa por ID para encontrar todos os eixos associados a cada matéria (ID único)
eixos_por_id = df_limpo.groupby('ID')['Eixo interpretativo'].unique().reset_index()

# 5. Dicionário para contar os pares de eixos
pares_contagem = defaultdict(int)

for eixos in eixos_por_id['Eixo interpretativo']:
    # Ordena os eixos para que (A, B) seja igual a (B, A)
    eixos_unicos = sorted(list(set(eixos)))
    if len(eixos_unicos) > 1:
        # Gera todas as combinações de pares possíveis na mesma matéria
        for eixo1, eixo2 in combinations(eixos_unicos, 2):
            pares_contagem[(eixo1, eixo2)] += 1

# 6. Transforma o dicionário em um DataFrame para exibir como tabela
dados_tabela = []
for (e1, e2), freq in pares_contagem.items():
    dados_tabela.append({
        'Eixo 1': e1,
        'Eixo 2': e2,
        'Matérias em Comum (IDs Únicos)': freq
    })

tabela_cruzamento = pd.DataFrame(dados_tabela)

# 7. Ordena do par mais frequente para o menos frequente
if not tabela_cruzamento.empty:
    tabela_cruzamento = tabela_cruzamento.sort_values(by='Matérias em Comum (IDs Únicos)', ascending=False)
    print("--- Tabela de Coocorrência entre Eixos Interpretativos ---")
    print(tabela_cruzamento.to_string(index=False))

    # 8. Filtro específico para o texto da dissertação (Ameaça à liberdade de expressão x Autoritarismo judicial)
    alvo_1 = "Ameaça à liberdade de expressão"
    alvo_2 = "Autoritarismo judicial"

    resultado_especifico = tabela_cruzamento[
        ((tabela_cruzamento['Eixo 1'] == alvo_1) & (tabela_cruzamento['Eixo 2'] == alvo_2)) |
        ((tabela_cruzamento['Eixo 1'] == alvo_2) & (tabela_cruzamento['Eixo 2'] == alvo_1))
    ]

    print("\n" + "="*50)
    if not resultado_especifico.empty:
        freq_encontrada = resultado_especifico['Matérias em Comum (IDs Únicos)'].values[0]
        print(f"-> O número de matérias em comum (coocorrência) entre:")
        print(f"   '{alvo_1}' e '{alvo_2}' é: {freq_encontrada}")
    else:
        print(f"-> Não foram encontradas matérias com coocorrência simultânea entre '{alvo_1}' e '{alvo_2}'.")
    print("="*50)
else:
    print("Nenhum cruzamento de múltiplos eixos por ID foi encontrado.")
```

```python
import pandas as pd
from itertools import combinations
from collections import defaultdict

# 1. Lê o CSV
df = pd.read_csv('/content/Relação das características.csv', header=1)
df_limpo = df.dropna(subset=['ID', 'Eixo interpretativo']).copy()
df_limpo['Eixo interpretativo'] = df_limpo['Eixo interpretativo'].str.strip()

# 2. Agrupa por ID
eixos_por_id = df_limpo.groupby('ID')['Eixo interpretativo'].unique().reset_index()

# 3. Conta os pares (garantindo ida e volta para preencher a matriz espelhada)
pares_contagem = defaultdict(int)

for eixos in eixos_por_id['Eixo interpretativo']:
    eixos_unicos = sorted(list(set(eixos)))
    if len(eixos_unicos) > 1:
        for eixo1, eixo2 in combinations(eixos_unicos, 2):
            pares_contagem[(eixo1, eixo2)] += 1
            pares_contagem[(eixo2, eixo1)] += 1  # Espelha para a matriz ficar completa

# 4. Cria lista de dados para a matriz
dados_tabela = []
for (e1, e2), freq in pares_contagem.items():
    dados_tabela.append({
        'Eixo 1': e1,
        'Eixo 2': e2,
        'Freq': freq
    })

tabela_cruzamento = pd.DataFrame(dados_tabela)

if not tabela_cruzamento.empty:
    # 5. Cria a matriz simétrica perfeita
    matriz_simetrica = tabela_cruzamento.pivot(
        index='Eixo 1',
        columns='Eixo 2',
        values='Freq'
    ).fillna(0).astype(int)

    print("="*100)
    print("--- MATRIZ SIMÉTRICA PERFEITA DE COOCORRÊNCIA ---")
    print("="*100)
    print(matriz_simetrica)
    print("="*100)
else:
    print("Nenhum cruzamento encontrado.")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from itertools import combinations
from collections import defaultdict

# 1. Lê o CSV
df = pd.read_csv('/content/Relação das características.csv', header=1)
df_limpo = df.dropna(subset=['ID', 'Eixo interpretativo']).copy()
df_limpo['Eixo interpretativo'] = df_limpo['Eixo interpretativo'].str.strip()

# 2. Agrupa por ID
eixos_por_id = df_limpo.groupby('ID')['Eixo interpretativo'].unique().reset_index()

# 3. Conta os pares (garantindo ida e volta para preencher a matriz espelhada)
pares_contagem = defaultdict(int)

for eixos in eixos_por_id['Eixo interpretativo']:
    eixos_unicos = sorted(list(set(eixos)))
    if len(eixos_unicos) > 1:
        for eixo1, eixo2 in combinations(eixos_unicos, 2):
            pares_contagem[(eixo1, eixo2)] += 1
            pares_contagem[(eixo2, eixo1)] += 1

# 4. Cria lista de dados para a matriz
dados_tabela = []
for (e1, e2), freq in pares_contagem.items():
    dados_tabela.append({
        'Eixo 1': e1,
        'Eixo 2': e2,
        'Freq': freq
    })

tabela_cruzamento = pd.DataFrame(dados_tabela)

if not tabela_cruzamento.empty:
    # 5. Cria a matriz simétrica
    matriz_simetrica = tabela_cruzamento.pivot(
        index='Eixo 1',
        columns='Eixo 2',
        values='Freq'
    ).fillna(0).astype(int)

    # 6. Ordem exata solicitada
    ordem_desejada = [
        "Ameaça à liberdade de expressão",
        "Autoritarismo judicial",
        "Defesa da Soberania",
        "Descrição de fatos",
        "Instabilidade política",
        "Proteção do EDD",
        "Responsabilização das plataformas",
        "Tentativa de interferência estrangeira"
    ]

    # Reindexa as linhas e as colunas na ordem exata
    matriz_simetrica = matriz_simetrica.reindex(index=ordem_desejada, columns=ordem_desejada, fill_value=0)

    # 7. Configura e gera o Heatmap (Mapa de Calor)
    plt.figure(figsize=(10, 8))
    sns.set_theme(style="white")

    # Desenha o heatmap com números nas células (annot=True) e cores intuitivas (Blues)
    ax = sns.heatmap(
        matriz_simetrica,
        annot=True,
        fmt="d",
        cmap="Blues",
        cbar=True,
        linewidths=.5,
        square=True
    )

    plt.title("Mapa de Calor: Coocorrência entre Eixos Interpretativos", fontsize=14, pad=20, fontweight='bold')
    plt.xlabel("Eixos Interpretativos", fontsize=11, labelpad=10)
    plt.ylabel("Eixos Interpretativos", fontsize=11, labelpad=10)

    # Roticiona os nomes dos eixos para ficarem legíveis
    plt.xticks(rotation=45, ha='right', fontsize=9)
    plt.yticks(rotation=0, fontsize=9)

    plt.tight_layout()
    plt.show()
else:
    print("Nenhum cruzamento encontrado.")
```
